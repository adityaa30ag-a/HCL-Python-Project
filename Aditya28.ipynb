```python
import pandas as pd
import numpy as np


class SalesRevenueAnalyzer:

    def __init__(self, dataframe):
        self.df = dataframe.copy()
        self.prepare_data()

    # Prepare data
    def prepare_data(self):

        self.df["Date"] = pd.to_datetime(
            self.df["Date"], errors="coerce"
        )

        self.df["Quantity"] = pd.to_numeric(
            self.df["Quantity"], errors="coerce"
        )

        self.df["UnitPrice"] = pd.to_numeric(
            self.df["UnitPrice"], errors="coerce"
        )

        self.df = self.df.dropna(
            subset=[
                "Date", "Product", "Category",
                "Quantity", "UnitPrice", "Region"
            ]
        )

        self.df = self.df[
            (self.df["Quantity"] > 0) &
            (self.df["UnitPrice"] > 0)
        ].copy()

        self.df["Revenue"] = (
            self.df["Quantity"] * self.df["UnitPrice"]
        )

        self.df["Month"] = (
            self.df["Date"].dt.to_period("M").astype(str)
        )

    # Show data
    def show_data(self):

        print("\nSALES DATA")

        print("Number of records:", len(self.df))

        print("Number of products:",
              self.df["Product"].nunique())

        print("\nFirst five records:")

        print(self.df.head().to_string(index=False))

    # Total revenue
    def total_revenue(self):

        return np.sum(self.df["Revenue"].to_numpy())

    # Revenue statistics
    def revenue_statistics(self):

        revenue = self.df["Revenue"].to_numpy()

        print("\nREVENUE STATISTICS")

        print("Total revenue:", np.sum(revenue))

        print("Average revenue:", np.mean(revenue))

        print("Highest revenue:", np.max(revenue))

        print("Lowest revenue:", np.min(revenue))

    # Product-wise revenue
    def product_revenue(self):

        return (
            self.df.groupby("Product")["Revenue"]
            .sum()
            .sort_values(ascending=False)
        )

    # Category-wise revenue
    def category_revenue(self):

        return (
            self.df.groupby("Category")["Revenue"]
            .sum()
            .sort_values(ascending=False)
        )

    # Region-wise revenue
    def region_revenue(self):

        return (
            self.df.groupby("Region")["Revenue"]
            .sum()
            .sort_values(ascending=False)
        )

    # Monthly revenue
    def monthly_revenue(self):

        return (
            self.df.groupby("Month")["Revenue"]
            .sum()
            .sort_index()
        )

    # Best selling products
    def best_selling_products(self):

        return (
            self.df.groupby("Product")["Quantity"]
            .sum()
            .sort_values(ascending=False)
        )

    # Revenue contribution
    def revenue_contribution(self):

        category_sales = self.category_revenue()

        total = np.sum(category_sales.to_numpy())

        percentage = (
            category_sales / total
        ) * 100

        return percentage.round(2).sort_values(
            ascending=False
        )

    # Average order value
    def average_order_value(self):

        order_revenue = (
            self.df.groupby("OrderID")["Revenue"]
            .sum()
        )

        return np.mean(order_revenue.to_numpy())

    # High revenue orders
    def high_revenue_orders(self):

        result = self.df[
            self.df["Revenue"] >= 1000
        ]

        return result[
            ["OrderID", "Product", "Revenue"]
        ]

    # Save reports
    def save_reports(self):

        self.product_revenue().to_csv(
            "product_revenue.csv"
        )

        self.category_revenue().to_csv(
            "category_revenue.csv"
        )

        self.region_revenue().to_csv(
            "region_revenue.csv"
        )

        self.monthly_revenue().to_csv(
            "monthly_revenue.csv"
        )

        self.revenue_contribution().to_csv(
            "revenue_contribution.csv"
        )

        print("\nReports saved successfully!")


# Load data
def load_data():

    try:

        dataframe = pd.read_csv("retail_sales.csv")

        print("CSV file loaded successfully.")

    except FileNotFoundError:

        print("CSV file not found.")

        print("Please upload retail_sales.csv")

        return None

    return dataframe


# Main program
def main():

    dataframe = load_data()

    if dataframe is None:
        return

    analyzer = SalesRevenueAnalyzer(dataframe)

    analyzer.show_data()

    print("\nTotal Revenue:",
          analyzer.total_revenue())

    analyzer.revenue_statistics()

    print("\nProduct-wise Revenue")
    print(analyzer.product_revenue())

    print("\nCategory-wise Revenue")
    print(analyzer.category_revenue())

    print("\nRegion-wise Revenue")
    print(analyzer.region_revenue())

    print("\nMonthly Revenue")
    print(analyzer.monthly_revenue())

    print("\nBest Selling Products")
    print(analyzer.best_selling_products())

    print("\nRevenue Contribution")
    print(analyzer.revenue_contribution())

    print("\nAverage Order Value:",
          round(analyzer.average_order_value(), 2))

    print("\nHigh Revenue Orders")
    print(analyzer.high_revenue_orders())

    analyzer.save_reports()


if __name__ == "__main__":
    main()
```
